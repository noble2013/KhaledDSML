علم البيانات وتعلم الآلة — السنة الثالثة اقتصاد وتسيير المؤسسات

> البيانات هنا اصطناعية لأغراض تعليمية. يمكن تشغيل الدفتر محلياً بعد تثبيت KhaledDSML.

## إنشاء بيانات معاملات

```python
from khaled_dsml import generate_transactions, quality_report, missingness_report
df = generate_transactions(2000)
print(quality_report(df))
missingness_report(df)
```

## إحصاء وصفي وقيم شاذة

```python
from khaled_dsml import descriptive_summary, iqr_outliers, fill_missing_numeric
descriptive_summary(df)
```

## فحص الشواذ وتعويض النواقص

```python
print("IQR outliers:", iqr_outliers(df,"montant_da").sum())
df_clean = fill_missing_numeric(df, strategy="median")
print("Remaining missing:", df_clean.montant_da.isna().sum())
```

## رسم القيم المفقودة

```python
from khaled_dsml.plots import plot_missingness
fig, ax = plot_missingness(df)
```
