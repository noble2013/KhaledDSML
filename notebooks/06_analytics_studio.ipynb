## 2. نتائج علمية حقيقية من بيانات تعليمية اصطناعية

تُحسب قيم الاختبارات بواسطة المكتبة، ولا تُكتب في اللوحة يدويًا.

</div>

```python
import khaled_dsml
from khaled_dsml import (
    mann_kendall_trend_test, pettitt_change_point_test,
    forecast_accuracy_report, display_bilingual_results,
    display_analysis_dashboard, save_analysis_dashboard,
)
print("Development version:", khaled_dsml.__version__)

observed_sales = [100, 106, 111, 119, 127, 136, 141, 153, 160, 172, 181, 196]
trend = mann_kendall_trend_test(observed_sales)
change = pettitt_change_point_test([12] * 20 + [30] * 20)
forecast = forecast_accuracy_report(
    actual=[205, 215, 225],
    prediction=[203, 218, 222],
    training=[100, 110, 120, 130, 140, 150]
)
```

<div dir="rtl" style="text-align:right;font:600 18px Tahoma,Arial,sans-serif">

## 3. لوحة الدراسة الإبداعية

الافتراضي: **Aurora**. يمكنك تعديل `theme` إلى `ocean` أو `amethyst`. الرسم مستمد من `observed_sales` فقط.

</div>

```python
display_analysis_dashboard(
    [trend, change, forecast],
    title_ar="مرصد الأداء الاقتصادي: دراسة المبيعات",
    title_en="ECONOMIC PERFORMANCE · INSIGHTS",
    theme="aurora",
    font_size=21,
    chart_data=observed_sales,
)
```

<div dir="rtl" style="text-align:right;font:600 18px Tahoma,Arial,sans-serif">

## 4. التبديل بين الأنماط

اعرض إحدى نتائج الاختبارات في نمط آخر، أو احتفظ بالإخراج النصي الكلاسيكي عبر `style="plain"`.

</div>

```python
display_bilingual_results(
    [trend, change],
    theme="amethyst",
    chart_data=observed_sales,
)
# بديل أبسط:
# display_bilingual_results([trend, change], style="plain", font_size=24)
```

## 5. حفظ التقرير النهائي وطباعته

ينشئ HTML مستقلًا يعمل دون إنترنت. افتحه في المتصفح، ثم استخدم **Print → Save as PDF** للطباعة.

</div>

```python
report_file = save_analysis_dashboard(
    [trend, change, forecast], "economic_studio_ar_en.html",
    title_ar="تقرير الإحصاء الاقتصادي",
    title_en="ECONOMIC ANALYSIS REPORT",
    theme="ocean", chart_data=observed_sales,
)
print("Saved:", report_file)
```
