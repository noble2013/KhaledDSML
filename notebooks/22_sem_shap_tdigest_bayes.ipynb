## 1 — SEM and measurement invariance | مؤشرات SEM وتكافؤ القياس

Indices and chi-square values must come from **already fitted compatible conventional ML SEM/CFA models**. / لا تقوم الأداة بملاءمة نموذج القياس.

```python
import numpy as np
from khaled_dsml import (
 sem_fit_indices, measurement_invariance_nested_ml,
 shap_explanation_stability, tdigest_quantile_summary,
 dirichlet_multinomial_bayes_factor,
 sprite_integer_constraint_witness, display_analysis_dashboard,
)
rng=np.random.default_rng(22)
display_analysis_dashboard(sem_fit_indices(65,40,250,60,180),theme="ocean")
display_analysis_dashboard(measurement_invariance_nested_ml(110,90,117,95),theme="amethyst")
```

## 2 — SHAP attribution stability | ثبات تفسيرات SHAP

Supply PRECOMPUTED SHAP values for identically aligned held-out rows, features, baselines and output scales. / لا تحسب هذه الدالة SHAP مباشرة.

```python
same_holdout_explanations=np.stack([
   rng.normal(size=(40,4))*[.1,.3,.7,1.6]
   for _ in range(4)
])
shap_result=shap_explanation_stability(
    same_holdout_explanations,feature_names=["income","age","volume","price"])
display_analysis_dashboard(shap_result,theme="aurora")
```

## 3 — t-digest | التقدير التقريبي لكميات البيانات الضخمة

The approximate quantiles are NOT exact population quantiles. / التقدير تقريبي وليس اختبارًا إحصائيًا.

```python
values=rng.normal(size=1200)
digest_result=tdigest_quantile_summary(values,probabilities=(.01,.1,.5,.9,.99))
display_analysis_dashboard(digest_result,theme="ocean")
```

## 4 — Exact Bayes factor | عامل بايز الدقيق للفئات

Select the Dirichlet prior before seeing the counts. / الحساسية للتوزيع القبلي عالية في العينات الصغيرة.

```python
bayes=dirichlet_multinomial_bayes_factor(
    counts=[4,2,1],null_probabilities=[.3,.3,.4],
    prior_concentration=[1,1,1])
display_analysis_dashboard(bayes,theme="amethyst")
```
