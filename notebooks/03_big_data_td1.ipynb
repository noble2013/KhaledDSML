ماستر 2 اقتصاد كمي — البيانات الضخمة. المعالجة باستخدام chunksize لا تحتاج تحميل الملف كاملاً؛ التوليد هنا يتم في الذاكرة.

> البيانات هنا اصطناعية لأغراض تعليمية. يمكن تشغيل الدفتر محلياً بعد تثبيت KhaledDSML.

## توليد مليون معاملة وحفظها

```python
from pathlib import Path
from khaled_dsml import generate_transactions,quality_report,optimize_memory
rows = 1_000_000  # قلل القيمة إذا كانت RAM صغيرة
df = generate_transactions(rows,missing_rate=.02)
print("Before",quality_report(df)["memory_mb"])
print("After",quality_report(optimize_memory(df))["memory_mb"])
filename=Path("transactions.csv")
df.to_csv(filename,index=False)
del df
```

## التجميع بالولاية

```python
from khaled_dsml import csv_chunk_summary
summary=csv_chunk_summary(filename,chunksize=200_000)
summary
```

## نسب النقص حسب الأعمدة

```python
from khaled_dsml import csv_chunk_missing
csv_chunk_missing(filename,chunksize=200_000)
```

## التحليل الشهري

```python
from khaled_dsml import csv_chunk_monthly
monthly=csv_chunk_monthly(filename,chunksize=200_000)
monthly
```
