## 1 — Phillips–Perron | اختبار جذر الوحدة

Null is a **unit root**; failure to reject does not confirm nonstationarity. A deterministic trend and bandwidth should be specified based on the actual design.

```python
import numpy as np
from khaled_dsml import (
  phillips_perron_unit_root, grimmer_variance_consistency,
  pcurve_right_skew_screen, display_analysis_dashboard,
)
rng=np.random.default_rng(2026)
walk=np.cumsum(rng.normal(size=250))
pp=phillips_perron_unit_root(walk,trend="c",test_type="tau")
display_analysis_dashboard(pp,theme="ocean")
```

## 2 — GRIMMER | اختبار الاتساق الحسابي

Mean and **sample SD** (ddof=1) from bounded integer responses, with ROUND_HALF_UP. If a pair is possible it does **not** establish that the original data are accurate. / نتائج حسابية لا تكشف تزويرًا.

```python
valid=grimmer_variance_consistency(
  "2.00","1.41",sample_size=2,minimum_score=1,maximum_score=3
)
invalid=grimmer_variance_consistency(
  "2.00","1.42",sample_size=2,minimum_score=1,maximum_score=3
)
display_analysis_dashboard(valid,theme="amethyst")
display_analysis_dashboard(invalid,theme="ocean")
```

## 3 — Half-p-curve | نصف منحنى قيم p الدالة

Not full p-curve analysis and **not a publication-bias diagnosis**. Valid under independent exact p-values conditional-uniform under a prespecified null. / يتطلب استقلالية الاختبارات وقيم p دقيقة لا منتقاة.

```python
pvalues=[.001,.002,.009,.018,.021,.026,.032,.044]
screen=pcurve_right_skew_screen(pvalues)
display_analysis_dashboard(screen,theme="aurora")
```

## Scientific limitations | حدود تفسير النتائج

- PP needs appropriately sampled time series, a prespecified deterministic trend and bandwidth, and review of structural breaks.
- GRIMMER only checks logical integer moment feasibility. Rounding conventions, missing items and score rules matter.
- A half-p-curve binomial test is NOT the full p-curve suite; selection and multiple testing can invalidate its p-value.
- 248 candidate-mapped catalogue rows are NOT proof of 248 scientifically completed methods.
