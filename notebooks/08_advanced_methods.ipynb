## 2. فحوص الافتراضات والتكافؤ

نتائج محسوبة من بيانات مصطنعة، دون قيم احتمالية مختلقة.

</div>

```python
import numpy as np
import khaled_dsml as k
from IPython.display import display

print("KhaledDSML:", k.__version__)
rng=np.random.default_rng(42)
a=rng.normal(size=(80,3))
b=rng.normal(size=(90,3))
covariance=k.box_m_covariance_equality(a,b)
mardia=k.mardia_multivariate_normality(a)
paired_a=rng.normal(loc=10,scale=.2,size=90)
paired_b=rng.normal(loc=10,scale=.2,size=90)
tost=k.welch_tost_equivalence(paired_a,paired_b,margin=.5)

k.display_analysis_dashboard(
    [covariance,mardia,tost],
    title_ar="مختبر فحص الافتراضات والتكافؤ",
    title_en="MULTIVARIATE ASSUMPTIONS · EQUIVALENCE",
    theme="aurora",font_size=21,
)
```

## 3. موثوقية المحكّمين

لا تخلط بين Alpha كريبندورف ومعامل الدلالة `alpha=0.05`.

</div>

```python
agreement=k.krippendorff_nominal_alpha([
    ["نعم","نعم",None],
    ["لا","لا","لا"],
    ["نعم","لا","نعم"],
    ["لا",None,"لا"],
])
k.display_analysis_dashboard(
    agreement,
    title_ar="مراجعة اتّفاق المحكّمين",
    title_en="INTER-RATER RELIABILITY",
    theme="amethyst",
)
```

<div dir="rtl" style="text-align:right">

## 4. مدونة عربية: التنوع المعجمي وقانون Zipf

تُعرض الجداول المشتقة فعليًا من الكلمات، والمقارنات الاحتمالية تتطلب تصحيحًا عند تعدد الاختبارات.

</div>

```python
corpus = (["الاقتصاد"]*45 + ["البحث"]*30 + ["البيانات"]*22
          + ["الجامعة"]*15 + ["التعليم"]*9 + ["الإحصاء"]*5 + ["الوادي"]*2)
lexicon=k.lexical_diversity(corpus)
zipf=k.zipf_frequency_analysis(corpus)
k.display_analysis_dashboard(
    [lexicon,zipf],
    title_ar="مرصد المدونة اللغوية العربية",
    title_en="ARABIC CORPUS · LEXICAL STATISTICS",
    theme="ocean",
)
```
