## 1. ANCOVA | تحليل التغاير

Use independent cases. Check homogeneity of slopes before interpreting the common-slope adjusted means. / تحقق من تجانس الميول قبل اعتماد المتوسطات المعدلة.

```python
import numpy as np
import pandas as pd
from khaled_dsml import (
    ancova_one_factor, manova_one_factor,
    repeated_measures_one_factor, display_analysis_dashboard,
)
rng = np.random.default_rng(1426)
groups = ["A"]*30 + ["B"]*30 + ["C"]*30
cov = rng.normal(10, 1.5, size=90)
effect = np.repeat([0., 1.0, 1.8], 30)
score = 5 + .6*cov + effect + rng.normal(0,.7,size=90)
ancova = ancova_one_factor(score, groups, cov)
display_analysis_dashboard(ancova,theme="ocean")
print("Slope homogeneity p:",ancova["slope_homogeneity_p_value"])
```

## 2. MANOVA | تحليل التباين المتعدد

Several continuous outcomes measured once per independent subject. The table lists four related multivariate statistics, not independent discoveries. / عدة مخرجات على حالات مستقلة.

```python
outcomes = rng.multivariate_normal(
    [10., 5., 2.],
    [[1., .25, .1],[.25,1.,.2],[.1,.2,1.]],
    size=90
)
outcomes += np.repeat([[0,0,0],[.5,-.2,.4],[1.1,.3,-.2]], 30, axis=0)
manova = manova_one_factor(outcomes,groups)
display_analysis_dashboard(manova,theme="amethyst")
```

## 3. Repeated-measures ANOVA | القياسات المتكررة

Rows = subjects, columns = repeated occasions. No missing visits. The classical F p-value assumes **sphericity**; no GG/HF correction is applied. / الصفوف أشخاص، الأعمدة مناسبات قياس متكررة، دون قيم مفقودة.

```python
subject = rng.normal(0,1,size=(28,1))
visits = subject + rng.normal(0,.3,size=(28,3)) + [0.,.35,.9]
wide = pd.DataFrame(visits,columns=["before","during","after"])
repeated = repeated_measures_one_factor(wide)
display_analysis_dashboard(repeated,theme="aurora")
```

## Methodology / حدود الاستدلال العلمي

- ANCOVA assumes appropriately specified conditional mean, **parallel covariate slopes**, independent subjects and reliable residual diagnostics.
- MANOVA assumes independent subjects and meaningful joint outcomes, with full-rank pooled covariance.
- Repeated ANOVA is a complete balanced design and uses uncorrected F; evaluate sphericity or choose a mixed-effects analysis with specialist support.
- Statistical significance does **not** establish causal explanation. / الدلالة الإحصائية لا تعني إثبات السببية.
