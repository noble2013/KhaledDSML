## انحدار المبيعات

```python
import numpy as np
import pandas as pd
from khaled_dsml import split_data, make_regressor, regression_metrics
rng = np.random.default_rng(42)
X = pd.DataFrame({"price":rng.uniform(10,100,180),"wilaya":rng.choice(["Alger","Oran"],180)})
y = 1200 - 7*X.price + rng.normal(0,20,180)
Xt,Xv,yt,yv=split_data(X,y)
model=make_regressor(Xt,"ridge",alpha=1).fit(Xt,yt)
regression_metrics(yv,model.predict(Xv))
```

## التصنيف

```python
from sklearn.datasets import make_classification
from khaled_dsml import make_classifier, classification_metrics
X,y=make_classification(n_samples=200,n_features=5,random_state=5)
Xt,Xv,yt,yv=split_data(X,y,stratify=True)
model=make_classifier(Xt,"logistic").fit(Xt,yt)
classification_metrics(yv,model.predict(Xv))
```

## التحقق المتقاطع

```python
from khaled_dsml import cross_validation_summary
model=make_classifier(X,"random_forest",n_estimators=30)
cross_validation_summary(model,X,y,task="classification",folds=5)
```

## التجميع

```python
from khaled_dsml import cluster_data, pca_summary
result=cluster_data(X,n_clusters=3)
print("silhouette:",result.silhouette)
print(pca_summary(X,n_components=2)["explained_variance_ratio"])
```
