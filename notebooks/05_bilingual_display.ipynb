<div dir="rtl" style="text-align:right">## 2. توليد نتائج تعليمية | Generate educational test results</div>

```python
import khaled_dsml
from khaled_dsml import (
    mann_kendall_trend_test,
    pettitt_change_point_test,
    forecast_accuracy_report,
    display_bilingual_result,
    display_bilingual_results,
    save_bilingual_results,
)

print("KhaledDSML version:", khaled_dsml.__version__)

trend = mann_kendall_trend_test([10, 12, 13, 15, 18, 19, 21, 23, 25, 27, 30, 32])
change = pettitt_change_point_test([10] * 20 + [30] * 20)
forecast = forecast_accuracy_report(
    [100, 110, 120], [102, 108, 119], [50, 60, 70, 80, 90]
)
```

<div dir="rtl" style="text-align:right">## 3. عرض كل النتائج دفعة واحدة بالاتجاه الصحيح | Batch bilingual display</div>

```python
display_bilingual_results([trend, change, forecast], font_size=24)
```

<div dir="rtl" style="text-align:right">## 4. عرض نتيجة واحدة أو حفظ HTML للطباعة</div>

```python
display_bilingual_result(trend, font_size=24)
path = save_bilingual_results(
    [trend, change, forecast],
    "khaled_dsml_bilingual_results.html",
    font_size=24
)
print("Saved:", path)
```
