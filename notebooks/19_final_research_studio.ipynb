## 1 — Sequential decisions and CUPED | القرار المتتابع وضبط تباين التجارب

SPRT needs simple hypotheses chosen before any observations. CUPED needs a **valid randomized experiment** and a **pre-treatment** covariate. / يجب تحديد الفروض مسبقًا وعدم تفسير الاستدلال التقريبي باعتباره سببياً دون تخصيص عشوائي.

```python
import numpy as np
import pandas as pd
from khaled_dsml import (
  bernoulli_sprt, cuped_randomized_effect, nested_ols_wald_lr_score,
  two_sample_power_sensitivity, hopkins_cluster_tendency,
  sample_population_comparison, hyperloglog_cardinality,
  minhash_lsh_candidates, scientific_report_alerts,
  display_analysis_dashboard,
)
rng=np.random.default_rng(272)
sprt=bernoulli_sprt(rng.binomial(1,.7,120),p0=.35,p1=.65)
display_analysis_dashboard(sprt,theme="amethyst")
n=60
t=np.r_[np.zeros(n,dtype=int),np.ones(n,dtype=int)]
prior=rng.normal(size=n*2)
outcome=20+1.4*prior+1.2*t+rng.normal(size=n*2)
display_analysis_dashboard(cuped_randomized_effect(outcome,t,prior),theme="ocean")
```

## 2 — Wald/LR/LM and prospective power | اختبارات الانحدار وحساسية القدرة

All three chi-square statistics here assume nested Gaussian iid OLS and are **not** universal model tests. / للمقارنة بين نموذجين خطيين متداخلين بشروط محددة.

```python
n=100
x=rng.normal(size=n)
z=rng.normal(size=n)
y=1+.6*x+.7*z+rng.normal(size=n)
restricted=np.column_stack([np.ones(n),x])
full=np.column_stack([np.ones(n),x,z])
display_analysis_dashboard(nested_ols_wald_lr_score(y,restricted,full),theme="aurora")
display_analysis_dashboard(two_sample_power_sensitivity(45,target_power=.8),theme="ocean")
```

## 3 — Hopkins and sample representativeness | التجميع وتمثيل المجتمع

The Hopkins reference is box-uniform and feature scaling matters. KS and chi-square cannot prove that selection of the sample is unbiased. / القياس وتغاير المتغيرات مهمان، وعدم الرفض لا يثبت تمثيلية العينة.

```python
X=rng.normal(size=(100,3))
display_analysis_dashboard(hopkins_cluster_tendency(X,sample_size=30),theme="amethyst")
population=pd.DataFrame({"value":rng.normal(size=120),"segment":["A"]*60+["B"]*60})
sample=pd.DataFrame({"value":rng.normal(.3,size=90),"segment":["A"]*40+["B"]*50})
display_analysis_dashboard(sample_population_comparison(
    population,sample,numeric=["value"],categorical=["segment"]
),theme="ocean")
```

## 4 — Streaming sketches / أدوات البيانات الضخمة التقريبية

HyperLogLog is an **approximation** of unique counts. MinHash/LSH gives **candidates**, not guaranteed exact duplicates. / عدد فريد تقريبي وأزواج تشابه محتملة تحتاج تأكيدًا.

```python
display_analysis_dashboard(hyperloglog_cardinality(range(1000)),theme="aurora")
documents=[["data","ml","science"],["science","ml","data"],["history","arabic","poem"],["data","science"]]
display_analysis_dashboard(minhash_lsh_candidates(documents,bands=8),theme="amethyst")
```
